```python
transactions = [1200, 450, 800, 1500, 2300, 700, 100]
```

```python
# calculate total transaction value without using sum().
total = 0
for i in transactions:
  total = total + i
print(f'The total is: {total}')
```

```python
# From the same list, find the highest and lowest transaction without max() and min().
highest = transactions[0]
lowest = transactions[0]
for i in transactions:
  if i> highest:
    highest = i
  if i < lowest:
    lowest = i
print(f'The highest is: {highest}')
print(f'The Lowest is: {lowest}')
```

```python
# find the average temperature.
temperatures = [32, 35, 28, 40, 38, 31, 42]
total = 0
nos = len(temperatures)
for i in temperatures:
  total = total + i
avg = total / nos
print(f'The average is : {avg:.2f}')
```

```python
# Given student marks

marks = [78, 92, 45, 67, 88, 53, 99]

# count how many students scored

# 90+

# 75–89

# 50–74

# below 50

count_1 = 0
count_2 = 0
count_3 = 0
count_4 = 0
for i in marks:
  if i >= 90:
    count_1 = count_1 + 1
  elif i >= 75 and i <= 89:
    count_2 = count_2 + 1
  elif i >= 50 and i <= 74:
    count_3 = count_3 + 1
  elif i < 50:
    count_4 = count_4 + 1
print(f'90+: {count_1}')
print(f'75-89: {count_2}')
print(f'50+: {count_3}')
print(f'Below 50: {count_4}')
```

```python
#Count the frequency of every character in a string without using Counter.

#Example

#banana

#b -> 1

#a -> 3

#n -> 2
str = 'Sanjeet'
frequency = {}
for char in str:
  if char in frequency:
    frequency[char] +=1
  else:
    frequency[char] = 1
for char,count in frequency.items():
  print(f'{char} -> {count}')
```

```python
# Find the second-largest number in a list without using sort().
transactions = [1200, 450, 800, 1500, 2300, 700, 100]
largest = transactions[0]
second_largest = transactions[0]
for i in transactions:
  if i > largest:
    second_largest = largest
    largest = i
  elif i > second_largest and i != largest:
    second_largest = i

print(f'The largest is: {largest},The second largest is: {second_largest}')
```

```python
# Check whether a string is a palindrome using loops.
str = input('Enter the string: ')
reverse_str = ''
for char in str:
  reverse_str = char + reverse_str
if str == reverse_str:
  print(f'{str} is palindrom')
else:
  print(f'{str} is not palindrom')
```

```python
'''
Create a basic ATM simulation where a user can repeatedly

Check balance

Deposit money

Withdraw money

Exit
'''
balance = 10000.00

while True:
    print("\nATM Menu:")
    print("1. Check balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = int(input("Enter your choice from 1 to 4: "))

    if choice == 1:
        print(f"Your account balance is: {balance:.2f}")
    elif choice == 2:
        amount = float(input("Enter the amount to deposit: "))
        if amount > 0:
            balance += amount
            print(f"The updated balance is: {balance:.2f}")
        else:
            print("Enter a valid amount")
    elif choice == 3:
        amount = float(input("Enter the amount to withdraw: "))
        if 0 < amount <= balance:
            balance -= amount
            print(f"The updated balance is: {balance:.2f}")
        elif amount > balance:
            print("Insufficient balance to withdraw")
        else:
            print("Enter a valid amount")
    elif choice == 4:
        print("Thank you for using the ATM")
        break
    else:
        print("Enter a valid choice from 1 to 4")
```
